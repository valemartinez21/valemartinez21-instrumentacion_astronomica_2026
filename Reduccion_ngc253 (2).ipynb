```python
#Calculamos las dimensiones y los valores estadísticos del Master Dark
print("Dimensiones:", master_dark240s.shape)
print("Mínimo:", np.min(master_dark240s))
print("Máximo:", np.max(master_dark240s))
print("Promedio:", np.mean(master_dark240s))
```

```python
# Convertimos los Master Dark a float32
dark120 = master_dark120s.astype(np.float32)
dark240 = master_dark240s.astype(np.float32)
dark300 = master_dark300s.astype(np.float32)

# Tiempos de exposición
t1 = 120
t2 = 240
t3 = 300

# Promedio de los tiempos
t_prom = np.mean([t1, t2, t3])

# Promedio de los Master Dark
dark_prom = (dark120 + dark240 + dark300) / 3

# Corriente oscura (pendiente de la recta)
dark_current = (
    (t1 - t_prom) * dark120 +
    (t2 - t_prom) * dark240 +
    (t3 - t_prom) * dark300
) / (
    (t1 - t_prom)**2 +
    (t2 - t_prom)**2 +
    (t3 - t_prom)**2
)

# Bias estimado (intercepto)
master_bias_estimado = dark_prom - dark_current * t_prom
```

```python
# Flats corregidos del filtro B

flats_b = [
    flatb1, flatb2, flatb3, flatb4, flatb5,
    flatb6, flatb7, flatb8, flatb9, flatb10
]

corrected_flats_b = [
    flat.astype(np.float32)
    - master_bias_estimado
    - dark_current * 0.5
    for flat in flats_b
]
# Normalizamos individualmente los flats B
normalized_flats_b = [
    flat / np.median(flat)
    for flat in corrected_flats_b
]

# Combinamos los flats B
master_flatb = np.median(normalized_flats_b, axis=0)

# Normalizamos el Master Flat B
master_flatb_norm = master_flatb / np.median(master_flatb)
```

```python
# Flats del filtro V
flats_v = [
    flatv1, flatv2, flatv3, flatv4, flatv5,
    flatv6, flatv7, flatv8, flatv9, flatv10
]

# Tiempos de exposición correspondientes
tiempos_v = [
    0.5, 0.5, 0.5, 0.5, 0.5, 0.5,
    1.02, 1.02, 1.02, 1.02
]

# Corregimos los flats V
corrected_flats_v = [
    flat.astype(np.float32)
    - master_bias_estimado
    - dark_current * tiempo
    for flat, tiempo in zip(flats_v, tiempos_v)
]
# Normalizamos individualmente los flats V
normalized_flats_v = [
    flat / np.median(flat)
    for flat in corrected_flats_v
]

# Combinamos los flats V
master_flatv = np.median(normalized_flats_v, axis=0)

# Normalizamos el Master Flat V
master_flatv_norm = master_flatv / np.median(master_flatv)
```
